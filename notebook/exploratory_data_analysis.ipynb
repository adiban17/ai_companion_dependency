#### Exploring the `age` feature

- The `age` denotes the age of the person in the survey

```python
# Range
max_age = df['age'].max()
min_age = df['age'].min()
mode_age = df['age'].mode()
print(f"The age varies from {min_age} - {max_age} with the most common age being {mode_age[0]}")
```

```python
# Visualizing the age-frequency distribution
plt.figure(figsize=(15, 6))
sns.histplot(data=df, x='age', bins=30, kde=False, color='teal')
plt.title('Frequency Distribution by Age')
plt.xlabel('Age of the Participants')
plt.ylabel('Number of Participants')
```

Insights:
- The maximum age is 44
- The minimum age is 15
- The most common age is 42

#### Exploring the `gender` feature

- `gender` tells us the gender of the participant

```python
# Count of unique genders
gender_counts = df['gender'].value_counts()
gender_counts
```

```python
# Visulaizing the composition by gender
plt.figure(figsize=(8, 8))
plt.title('Survey Composition by Gender')
gender_counts.plot(
    kind='pie', 
    autopct='%1.1f%%',       
    startangle=140,         
    colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99'], 
    explode=(0.05, 0, 0, 0)  
)
```

Insights:
- There are 4 gender categories
- The highest count is that of `male` closely followed by `female`
- A very small portion of people `Prefer not to say` 

#### Exploring the `occupation_status` column

- The `occupation_status` tell us whether the participant is a `Student`, `Unemployed`, `Freelancer`, '`Employed Full-Time` or `Employed Part-Time`

```python
occupations = df['occupation_status'].value_counts()
occupations
```

```python
# Visualizing the occupation by frequency counts
plt.figure(figsize=(10, 8))
my_colors = ['#FF9999', '#66B3FF', '#99FF99', '#FFCC99', '#C9CBFF']
plt.xlabel('Occupation Status')
plt.ylabel('Counts')
plt.title('Distribution of Participants by Occupation')
plt.bar(x=occupations.index, height=occupations.values, color=my_colors);
```

Insights:
- Most of the participants are students
- The least number of participants are Freelancers and Unemployed
